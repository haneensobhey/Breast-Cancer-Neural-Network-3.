```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/data.csv')
```

```python
df.head()
df.isna().sum()
```

```python
# del df['id']
x = df.drop('diagnosis', axis=1)
x=df.drop('id',axis=1)
y=df['diagnosis']
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# فصل الـ Features عن الـ Target
x = df.drop(columns=['id', 'diagnosis'])
y = df['diagnosis']

# تحويل الـ Target إلى أرقام
le = LabelEncoder()
y = le.fit_transform(y)

# تقسيم البيانات
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.1, random_state=42
)

# Scaling
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
print("y_test:", y_test.shape)
print("ypred:", np.shape(ypred))
```

```python
model=Sequential()
model.add(Dense(32,activation='relu',input_dim=30))
model.add(Dense(16,activation='relu'))
model.add(Dense(16,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
from tensorflow.keras.utils import plot_model

plot_model(
    model,
    show_shapes=True,
    show_layer_names=True
)
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
his=model.fit(x_train,y_train,epochs=100,batch_size=50,validation_data=(x_test,y_test))
```

```python
import matplotlib.pyplot as plt
plt.plot(his.history['loss'],label='train_loss')
plt.plot(his.history['val_loss'],label='val_loss')
plt.legend()
plt.show()
plt.plot(his.history['accuracy'],label='train_acc')
plt.plot(his.history['val_accuracy'],label='val_acc')
plt.legend()
plt.show()
```

```python
ypred = model.predict(x_test)

ypred = (ypred >= 0.5).astype(int)
ypred = ypred.reshape(-1)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
cm=confusion_matrix(y_test,ypred)
print(cm)
print(classification_report(y_test,ypred))
```
